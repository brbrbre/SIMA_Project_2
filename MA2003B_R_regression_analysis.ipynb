```python
import pandas as pd
import copy

# Listas que ya tienes
invalid_flags = ['p', 'c', 'd', 'b', 'm', 'l', 'z', 'o', 's', 'r', 'e', 'a', 'f', 'h', 'n', 'x']
all_quant_vars = ['CO', 'NO', 'NO2', 'NOX', 'O3', 'PM10', 'PM2.5', 'SO2', 'PRS', 'RAINF', 'RH', 'SR', 'TOUT', 'WSR', 'WDR']
key_vars = ['NO2', 'PM2.5', 'SO2']  # variables importantes por defecto

def nulls_by_station(dataset: dict, dataset_name: str, prefer_key_vars: bool = True):
    """
    Retorna y imprime un dict { estacion: total_nulos }.
    - Si prefer_key_vars=True usa key_vars si existen; si no, usa all_quant_vars presentes.
    - No modifica el dataset original.
    """
    # Copia superficial (los DataFrames no se tocan) para iterar seguro
    ds_keys = [k for k in dataset.keys() if k != 'CATÁLOGO']
    resultado = {}

    for station in ds_keys:
        df = dataset[station]
        if not isinstance(df, pd.DataFrame) or df.empty:
            resultado[station] = 0
            continue

        # Elegir columnas: primero key_vars si hay, si no all_quant_vars presentes
        cols_key = [c for c in key_vars if c in df.columns]
        cols_all = [c for c in all_quant_vars if c in df.columns]
        cols = cols_key if (prefer_key_vars and len(cols_key) > 0) else cols_all

        if len(cols) == 0:
            # Si no hay columnas cuantitativas conocidas, cuenta 0 nulos
            resultado[station] = 0
            continue

        # Convertir a numérico sin modificar el original
        df_numeric = df.copy()
        for c in cols:
            df_numeric[c] = pd.to_numeric(df_numeric[c], errors='coerce')

        # Sumar nulos solo en columnas seleccionadas
        total_nulos = int(df_numeric[cols].isna().sum().sum())
        resultado[station] = total_nulos

    # Imprimir ordenado
    print("="*70)
    print(f"Nulos por estación — {dataset_name}")
    print("-"*70)
    for est, nulos in sorted(resultado.items(), key=lambda x: x[1], reverse=True):
        print(f"{est:25s}  ->  {nulos} nulos")

    return resultado

# Ejemplos de uso con tus bases:
nulos_2324 = nulls_by_station(db_2022_2024_variablesImportantes, "2022–2024", prefer_key_vars=True)
```

```python
import pandas as pd
import numpy as np

# Definir las variables de gases
gases = ['CO', 'NO', 'NO2', 'NOX', 'O3', 'SO2', 'PM10', 'PM2.5']

# Función para calcular y mostrar el ranking de estaciones por contaminación
def rank_contaminacion_por_estacion(dataset: dict, dataset_name: str):
    """
    Calcula un índice de contaminación por estación basado en la media normalizada de los gases.
    Muestra una lista ordenada de estaciones de más a menos contaminadas.
    - dataset: Diccionario con dataframes por estación.
    - dataset_name: Nombre del dataset (ej. '2023-2024').
    """
    resultados = []
    
    for station in dataset:
        df = dataset[station]
        if not isinstance(df, pd.DataFrame) or df.empty:
            print(f"Estación {station} no tiene datos válidos. Saltando.")
            continue
        
        # Filtrar solo columnas de gases presentes
        gases_presentes = [gas for gas in gases if gas in df.columns]
        if len(gases_presentes) == 0:
            print(f"Estación {station} no tiene gases disponibles. Saltando.")
            continue
        
        # Calcular la media de cada gas (ignorando NaN)
        medias = df[gases_presentes].mean()
        
        # Normalizar las medias (escalado Min-Max para comparar gases con unidades diferentes)
        medias_normalizadas = {}
        for gas in gases_presentes:
            min_val = df[gas].min()
            max_val = df[gas].max()
            if max_val > min_val:  # Evitar división por cero
                medias_normalizadas[gas] = (medias[gas] - min_val) / (max_val - min_val)
            else:
                medias_normalizadas[gas] = 0  # Si no hay variación, asignar 0
        
        # Calcular índice de contaminación (promedio de medias normalizadas)
        indice_contaminacion = np.mean(list(medias_normalizadas.values()))
        
        resultados.append({
            'Estación': station,
            'Índice de Contaminación': indice_contaminacion,
            'Medias': medias.to_dict()  # Guardar medias originales para referencia
        })
    
    # Convertir a DataFrame y ordenar de mayor a menor contaminación
    df_resultados = pd.DataFrame(resultados)
    if df_resultados.empty:
        print(f"No hay datos válidos para {dataset_name}.")
        return
    
    df_resultados = df_resultados.sort_values(by='Índice de Contaminación', ascending=False)
    
    # Imprimir ranking
    print("="*70)
    print(f"Ranking de Estaciones por Contaminación - {dataset_name}")
    print("-"*70)
    print(f"{'Rank':<5} {'Estación':<25} {'Índice de Contaminación':<25} {'Medias de Gases'}")
    print("-"*70)
    for idx, row in df_resultados.iterrows():
        medias_str = ', '.join([f"{k}: {v:.2f}" for k, v in row['Medias'].items()])
        print(f"{idx+1:<5} {row['Estación']:<25} {row['Índice de Contaminación']:.4f}              {medias_str}")
    
    return df_resultados

# Generar rankings para ambos datasets

rank_2324 = rank_contaminacion_por_estacion(db_2022_2024_variablesImportantes, "2022-2024")
```

```python
def hourly_pollutants_df(data_dict):
    """
    Para cada estación en el diccionario:
    - Convierte la columna 'date' de epoch a datetime
    - Calcula el promedio por hora del día de los contaminantes
    - Grafica las 24 horas con el promedio de cada contaminante
    """
    mean_dfs = {}
    pollutants = ["CO", "NO", "NO2", "NOX", "O3", "PM10", "PM2.5", "SO2"]

    for station, df in data_dict.items():
        # Convertir la columna 'date' a datetime
        df['date'] = pd.to_datetime(df['date'], unit='s')

        # Crear columna con la hora del día
        df['Hora'] = df['date'].dt.hour

        # Filtrar solo contaminantes + Hora
        df_pollutants = df[pollutants + ['Hora', 'weekday']]

        #Eliminar días que no sean de semana
        df_pollutants_weekday = df_pollutants[df_pollutants['weekday'] < 5].copy()

        # Calcular promedios por hora
        hourly_means = df_pollutants_weekday.groupby('Hora').mean()
        mean_dfs[station] = hourly_means
    
    return mean_dfs
```

```python
import pandas as pd
import numpy as np
from scipy import stats

def analyze_pollutant_patterns(data_dict, confidence_level=0.95):
    """
    Analyze hourly pollutant patterns with confidence intervals
    to determine if observed patterns are statistically significant
    
    Statistical basis: 
    Confidence intervals help us understand the precision of our mean estimates.
    Narrow intervals suggest the mean is a reliable indicator of the typical value,
    while wide intervals indicate high variability that might make averages misleading.
    
    Source: Statistical methods based on Central Limit Theorem and 
    t-distribution for small sample sizes (Gosset, 1908; Fisher, 1925)
    """

    pollutants = ["NO", "NOX", "PM10"]
    results = {}
    
    for station, df in data_dict.items():
        print(f"\n=== Analyzing {station} ===")
        
        station_results = {}
        
        for poll in pollutants:
            # Calculate statistics for each hour
            hourly_stats = []
            for hour in range(24):
                hour_data = df[df['Hora'] == hour][poll]
                n = len(hour_data)
                
                if n > 0:
                    mean = np.mean(hour_data)
                    std = np.std(hour_data, ddof=1)  # Sample standard deviation
                    
                    # Calculate confidence interval
                    if n > 1:
                        se = std / np.sqrt(n)  # Standard error
                        t_value = stats.t.ppf((1 + confidence_level) / 2, n - 1)
                        ci_lower = mean - t_value * se
                        ci_upper = mean + t_value * se
                    else:
                        ci_lower = ci_upper = mean
                    
                    hourly_stats.append({
                        'hour': hour,
                        'mean': mean,
                        'ci_lower': ci_lower,
                        'ci_upper': ci_upper,
                        'n': n
                    })
            
            station_results[poll] = pd.DataFrame(hourly_stats)
            
            # Print results for this pollutant
            print(f"\n{poll} pattern analysis:")
            daily_range = station_results[poll]['mean'].max() - station_results[poll]['mean'].min()
            avg_ci_width = (station_results[poll]['ci_upper'] - station_results[poll]['ci_lower']).mean()
            
            print(f"  Daily concentration range: {daily_range:.3f}")
            print(f"  Average confidence interval width: {avg_ci_width:.3f}")
            
            # Assess pattern reliability
            if daily_range > 0:
                reliability_ratio = avg_ci_width / daily_range
                print(f"  Reliability ratio: {reliability_ratio:.3f}")
                
                if reliability_ratio < 0.3:
                    print("  ✅ Pattern is likely REAL (low variability relative to pattern strength)")
                elif reliability_ratio < 0.6:
                    print("  ⚠️ Pattern is MODERATELY reliable (some uncertainty)")
                else:
                    print("  ❌ Pattern may be MISLEADING (high variability masks true pattern)")
            else:
                print("  No discernible pattern (flat line)")
        
        results[station] = station_results
    
    return results

# Run the analysis
results = analyze_pollutant_patterns(f24_clean_weekday_v2)
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

def plot_NOX_confidence_intervals(data_dict, confidence_level=0.95):
    """
    Create visualization of NOX concentrations with confidence intervals for each station
    
    Statistical basis:
    Confidence intervals show the range where the true mean likely falls.
    Narrow intervals indicate precise estimates, while wide intervals suggest high variability.
    Overlapping intervals between hours suggest no statistically significant difference.
    """
    # Set up the plot
    n_stations = len(data_dict)
    fig, axes = plt.subplots(n_stations, 1, figsize=(12, 4 * n_stations))
    
    # If only one station, make sure axes is iterable
    if n_stations == 1:
        axes = [axes]
    
    # Colors for different stations
    colors = plt.cm.Set3(np.linspace(0, 1, n_stations))
    
    for idx, (station, df) in enumerate(data_dict.items()):
        ax = axes[idx]
        
        # Calculate hourly statistics
        hours = range(24)
        means = []
        ci_lowers = []
        ci_uppers = []
        
        for hour in hours:
            hour_data = df[df['Hora'] == hour]['NOX'].dropna()
            n = len(hour_data)
            
            if n > 0:
                mean = np.mean(hour_data)
                std = np.std(hour_data, ddof=1)  # Sample standard deviation
                
                if n > 1:
                    se = std / np.sqrt(n)  # Standard error
                    t_value = stats.t.ppf((1 + confidence_level) / 2, n - 1)
                    ci_lower = mean - t_value * se
                    ci_upper = mean + t_value * se
                else:
                    ci_lower = ci_upper = mean
            else:
                mean = ci_lower = ci_upper = np.nan
                
            means.append(mean)
            ci_lowers.append(ci_lower)
            ci_uppers.append(ci_upper)
        
        # Plot the data
        ax.plot(hours, means, 'o-', color=colors[idx], linewidth=2, markersize=6, label=f'{station}')
        ax.fill_between(hours, ci_lowers, ci_uppers, alpha=0.3, color=colors[idx])
        
        # Customize the plot
        ax.set_xlabel('Hora del día', fontsize=12)
        ax.set_ylabel('Concentración de NOx', fontsize=12)
        ax.set_title(f'{station} - Concentraciones de NOx con {int(confidence_level*100)}% de intervalo de confianza', fontsize=14)
        ax.grid(True, alpha=0.3)
        ax.legend()
        
        # Add statistical summary to the plot
        daily_range = np.nanmax(means) - np.nanmin(means)
        avg_ci_width = np.nanmean([u - l for l, u in zip(ci_lowers, ci_uppers)])
        
        if daily_range > 0:
            reliability_ratio = avg_ci_width / daily_range
            reliability_text = f"Confiabilidad del patrón: {reliability_ratio:.3f}\n"
            if reliability_ratio < 0.3:
                reliability_text += "Patrón fuerte"
            elif reliability_ratio < 0.6:
                reliability_text += "Patrón moderado"
            else:
                reliability_text += "Patrón débil (potencialmente engañoso)"
        else:
            reliability_text = "No hay un patrón visible"
            
        ax.text(0.02, 0.98, reliability_text, transform=ax.transAxes, verticalalignment='top',
                bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))
    
    plt.tight_layout()
    plt.savefig('NOX_confidence_intervals.png', dpi=300, bbox_inches='tight')
    plt.show()

# Prepare the NOX data
f24_clean_NOX = {}
for key, val in f24_clean_weekday_v2.items():
    f24_clean_NOX[key] = val[['date', 'NOX', 'weekday', 'Hora']]

# Generate the visualization
plot_NOX_confidence_intervals(f24_clean_NOX)
```

```python
import pandas as pd
import numpy as np
from scipy import stats
from datetime import datetime, timedelta

def campaign_analysis(data_dict, cutoff_date=pd.to_datetime("2024-12-30 23:00:00"), cont='NOX'):
    """
    Analyze the persistent impact of a campaign on pollutant levels
    with a focus on statistical significance testing
    
    Parameters:
    data_dict: Dictionary of DataFrames with pollution data
    cutoff_date: Date marking the end of the campaign
    cont: Pollutant to analyze (default: 'NOX')
    """
    
    for key, df in data_dict.items():
        print(f"\n=== Analysis for {key} ===")
        
        # Ensure date is datetime and sort
        df = df.copy()
        df['date'] = pd.to_datetime(df['date'])
        df = df.sort_values('date')
        
        # Define analysis periods
        pre_start = cutoff_date - timedelta(days=90)  # 3 months before
        post_end = cutoff_date + timedelta(days=90)   # 3 months after
        
        # Filter data to 3 months before and after campaign
        pre_campaign = df[(df['date'] >= pre_start) & (df['date'] <= cutoff_date)]
        post_campaign = df[(df['date'] > cutoff_date) & (df['date'] <= post_end)]
        
        print(f"Pre-campaign period: {pre_start.date()} to {cutoff_date.date()}")
        print(f"Post-campaign period: {cutoff_date.date()} to {post_end.date()}")
        print(f"Pre-campaign samples: {len(pre_campaign)}, Post-campaign samples: {len(post_campaign)}")
        
        # Check if we have enough data
        if len(pre_campaign) < 30 or len(post_campaign) < 30:
            print("Insufficient data for reliable analysis")
            continue
        
        # 1. Primary test: Mann-Whitney U test
        mw_stat, mw_p = stats.mannwhitneyu(pre_campaign[cont], post_campaign[cont])
        print(f"Mann-Whitney U test p-value: {mw_p:.6f}")
        
        # 2. Kolmogorov-Smirnov test for distributional differences
        ks_stat, ks_p = stats.ks_2samp(pre_campaign[cont], post_campaign[cont])
        print(f"Kolmogorov-Smirnov test p-value: {ks_p:.6f}")
        
        # 3. Bootstrap confidence interval for the median difference
        n_bootstraps = 5000
        boot_diffs = []
        for _ in range(n_bootstraps):
            pre_sample = np.random.choice(pre_campaign[cont], size=len(pre_campaign), replace=True)
            post_sample = np.random.choice(post_campaign[cont], size=len(post_campaign), replace=True)
            boot_diffs.append(np.median(post_sample) - np.median(pre_sample))
        
        median_diff = np.median(boot_diffs)
        ci_lower = np.percentile(boot_diffs, 2.5)
        ci_upper = np.percentile(boot_diffs, 97.5)
        print(f"Median difference (post - pre): {median_diff:.3f}")
        print(f"95% Confidence Interval: [{ci_lower:.3f}, {ci_upper:.3f}]")
        
        # 4. Weekly analysis to check persistence
        print("\nWeekly analysis of post-campaign period:")
        post_campaign['week'] = post_campaign['date'].dt.isocalendar().week
        weekly_means = post_campaign.groupby('week')[cont].mean()
        weekly_p_values = []
        
        for week in weekly_means.index:
            week_data = post_campaign[post_campaign['week'] == week][cont]
            if len(week_data) > 7:  # Ensure enough data points
                week_stat, week_p = stats.mannwhitneyu(pre_campaign[cont], week_data)
                weekly_p_values.append(week_p)
                print(f"Week {week}: p-value = {week_p:.4f}, mean = {weekly_means[week]:.3f}")
        
        # 5. Check for trend in post-campaign period
        if len(weekly_p_values) > 4:  # Need at least 4 weeks for trend analysis
            post_campaign['days_since_campaign'] = (post_campaign['date'] - cutoff_date).dt.days
            slope, intercept, r_value, p_value, std_err = stats.linregress(
                post_campaign['days_since_campaign'], post_campaign[cont]
            )
            print(f"Post-campaign trend: slope = {slope:.6f}, p-value = {p_value:.6f}")
            
            if p_value < 0.05:
                if slope < 0:
                    print("Significant decreasing trend in post-campaign period")
                else:
                    print("Significant increasing trend in post-campaign period")
            else:
                print("No significant trend in post-campaign period")
        
        # 6. Practical significance assessment
        pre_median = np.median(pre_campaign[cont])
        post_median = np.median(post_campaign[cont])
        percent_change = ((post_median - pre_median) / pre_median) * 100
        
        print(f"\nPractical significance:")
        print(f"Pre-campaign median: {pre_median:.3f}")
        print(f"Post-campaign median: {post_median:.3f}")
        print(f"Percent change: {percent_change:.2f}%")
        
        # 7. Overall conclusion
        print(f"\nConclusion for {key}:")
        if mw_p < 0.05:
            if median_diff < 0:
                print(f"Statistically significant decrease in {cont} after campaign")
                if all(p < 0.1 for p in weekly_p_values):  # Most weeks show significance
                    print("Decrease appears persistent across the post-campaign period")
                else:
                    print("Decrease may not be persistent across all weeks")
            else:
                print(f"Statistically significant increase in {cont} after campaign")
        else:
            print(f"No statistically significant change in {cont} after campaign")
            
            if ks_p < 0.05:
                print("However, distributional changes detected (KS test significant)")
        
        print("-" * 50)

# Example usage
if __name__ == "__main__":
    # Prepare the data (assuming f24_clean_weekday_v2 exists)
    f24_clean_NOx = {}
    for key, val in f24_clean_weekday_v2.items():
        f24_clean_NOx[key] = val[['date', 'NOX', 'weekday', 'Hora']].copy()
    
    # Set the actual cutoff date for your campaign
    campaign_end_date = pd.to_datetime("2023-04-05 23:00:00")
    
    # Run the analysis
    campaign_analysis(f24_clean_NOx, cutoff_date=campaign_end_date, cont='NOX')
```

```python
import seaborn as sns

def plot_campaign_relation(df, cutoff_date, cont="NOX", key="Dataset"):
    """
    Graficar la relación de contaminante antes y después de la campaña
    usando el mismo rango de tiempo que campaign_analysis.
    """
    # Copia y asegura el formato de fecha
    df = df.copy()
    df['date'] = pd.to_datetime(df['date'])
    df = df.sort_values('date')
    
    # Rango 90 días antes y después
    pre_start = cutoff_date - timedelta(days=90)
    post_end = cutoff_date + timedelta(days=90)
    df_range = df[(df['date'] >= pre_start) & (df['date'] <= post_end)]
    
    # Variable de periodo (pre / post)
    df_range['period'] = np.where(df_range['date'] <= cutoff_date, "Pre-campaign", "Post-campaign")
    
    # --- 1. Serie temporal con línea de corte ---
    plt.figure(figsize=(12,6))
    sns.lineplot(data=df_range, x="date", y=cont, hue="period", palette={"Pre-campaign":"blue", "Post-campaign":"red"})
    plt.axvline(cutoff_date, color="black", linestyle="--", label="Campaign End")
    plt.title(f"{key} - {cont} Levels Around Campaign")
    plt.ylabel(f"{cont} concentration")
    plt.legend()
    plt.tight_layout()
    plt.show()
    
    # --- 2. Comparación de distribuciones ---
    plt.figure(figsize=(8,6))
    sns.boxplot(data=df_range, x="period", y=cont, palette={"Pre-campaign":"blue", "Post-campaign":"red"})
    sns.swarmplot(data=df_range, x="period", y=cont, color=".25", size=3)  # puntos individuales
    plt.title(f"{key} - Distribution of {cont} Pre vs Post Campaign")
    plt.ylabel(f"{cont} concentration")
    plt.tight_layout()
    plt.show()
    
    # --- 3. Medias móviles para ver tendencia ---
    df_range = df_range.set_index("date")
    df_range[f"{cont}_rolling"] = df_range[cont].rolling(window=24*7, min_periods=10).mean()  # media móvil semanal
    
    plt.figure(figsize=(12,6))
    plt.plot(df_range.index, df_range[f"{cont}_rolling"], color="purple", label="Weekly rolling mean")
    plt.axvline(cutoff_date, color="black", linestyle="--", label="Campaign End")
    plt.title(f"{key} - Rolling Weekly Mean of {cont}")
    plt.ylabel(f"{cont} concentration")
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
from scipy import stats

def check_normality_simple(df, variables):

    results = []
    
    for var in variables:
        data = df[var]
        stat, p_value = stats.shapiro(data)
        skew = stats.skew(data)
        kurtosis = stats.kurtosis(data)
        
        results.append({
            'variable': var,
            'shapiro_p': p_value,
            'is_normal': p_value > 0.05,
            'skewness': skew,
            'kurtosis': kurtosis
        })
    
    return pd.DataFrame(results)
```

```python
from sklearn.preprocessing import PowerTransformer

def transform_pollution_data(df, variables):
    data_to_transform = df[variables].copy()
    
    transformer = PowerTransformer(method='yeo-johnson', standardize=True)
    transformed_data = transformer.fit_transform(data_to_transform)
    
    transformed_df = pd.DataFrame(transformed_data, 
                                columns=[f'{var}' for var in variables],
                                index=df.index)
    
    return transformed_df

trans_normal_df_24 = {}
for station, df in f24_clean_weekday.items():
    trans_normal_df_24[station] = transform_pollution_data(df, pollutants)
```

```python
def verify_transformation(original_dict, transformed_dict, variables):
    print(f"\nEfectividad de transformación:")
    
    for station_name in original_dict.keys():
        print(f"\n--- {station_name} ---")
        original_df = original_dict[station_name]
        transformed_df = transformed_dict[station_name]
        
        for i, var in enumerate(variables):
            _, p_original = stats.shapiro(original_df[var])
            _, p_transformed = stats.shapiro(transformed_df.iloc[:, i])
            print(f"{var}: {p_original:.6f} → {p_transformed:.6f}")

verify_transformation(f24_clean_weekday, trans_normal_df_24, pollutants)
```
